```python
df['fuel'].value_counts(dropna=False)
```

```python
df['fuel'] = df['fuel'].str[1].str.strip()
df['fuel'].value_counts(dropna=False)
```

```python
Gasoline = ['Gasoline', 'Gasoline (Particulate Filter)',  'Regular/Benzine 91', 'Regular/Benzine E10 91', 'Regular/Benzine 91 (Particulate Filter)', 'Super 95', 'Super 95 (Particulate Filter)', 'Super E10 95',
'Super E10 95 (Particulate Filter)', 'Super Plus 98', 'Super Plus E10 98']
Diesel = ['Diesel', 'Diesel (Particulate Filter)']
Gaseous = ['LPG', 'Liquid petroleum gas (LPG)', 'CNG', 'CNG (Particulate Filter)', 'Biogas', 'Domestic gas H']
Electric = ['Electric', 'Electric/Gasoline']
Others = ['Others', 'Others (Particulate Filter)']
```

```python
def fuel_type(x):
    if type(x) == str:
        x = x.split(' / ')
        if x[0] in Gasoline:
            return 'Gasoline'
        elif x[0] in Diesel:
            return 'Diesel'
        elif x[0] in Gaseous:
            return 'Gaseous'
        elif x[0] in Electric:
            return 'Electric'
        elif x[0] in Others:
            return 'Others'
    else:
        return np.nan
```

```python
df['fuel'] = df['fuel'].apply(fuel_type)
df['fuel'].value_counts(dropna=False)
```

```python
def split_comb(x):

    if type(x) == list and len(x) == 3:
        if x[0] != []: return x[0][0]
        else: return None

    if type(x) == list and len(x) == 5:
        return x[1]

    if type(x) == list and len(x) == 7:
        return x[1]

    else:
        return None
```

```python
def split_city(x):

    if type(x) == list and len(x) == 3:
        if x[1] != []: return x[1][0]
        else: return None

    if type(x) == list and len(x) == 5:
        return None

    if type(x) == list and len(x) == 7:
        return x[3]

    else:
        return None
```

```python
def split_country(x):

    if type(x) == list and len(x) == 3:
        if x[2] != []: return x[2][0]
        else: return None

    if type(x) == list and len(x) == 5:
        return None

    if type(x) == list and len(x) == 7:
        return x[5]

    else:
        return None
```

```python
df['consumption_comb'] = df['consumption'].apply(split_comb).str.extract(r'(\d+.\d|\d+)').astype("float")
df['consumption_comb'].value_counts(dropna=False)
```

```python
df['consumption_city'] = df['consumption'].apply(split_city).str.extract(r'(\d+.\d|\d+)').astype("float")
df['consumption_city'].value_counts(dropna=False)
```

```python
df['consumption_country'] = df['consumption'].apply(split_country).str.extract(r'(\d+.\d|\d+)').astype("float")
df['consumption_country'].value_counts(dropna=False)
```
