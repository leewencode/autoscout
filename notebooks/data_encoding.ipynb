### Numerical columns

### Categorical columns

```python
df_cat = df.select_dtypes(exclude=np.number)
df_cat.head()
col_cat_multi = ['entertainment_media', 'safety_security', 'comfort_convenience', 'extras']
df_cat_bin = df_cat.drop(columns=col_cat_multi)
col_cat_bin = df_cat_bin.columns
df_cat_multi = df_cat[col_cat_multi]
```

#### OneHot encoding for single cell single class categorical columns

```python
df_cat_bin_encode = pd.get_dummies(df_cat_bin, dtype=int, drop_first=True)
```

```python
df_cat_bin_encode.head()
```

```python
df_cat_bin_encode.shape
```

#### OneHot encoding for single cell single multiple classes categorical columns

```python
def ohe_multi(series):
    df_tmp = series.str.split(',').explode()
    df_tmp_ohe = pd.get_dummies(df_tmp, dtype=int)
    df_tmp_final = df_tmp_ohe.groupby(df_tmp.index).sum()
    return df_tmp_final
```

```python
df_cat_multi_encode = pd.DataFrame(index=df_cat_multi.index)

for col in col_cat_multi:
    df_cat_multi_tmp = ohe_multi(df_cat_multi[col])
    df_cat_multi_encode = pd.concat([df_cat_multi_encode, df_cat_multi_tmp], axis=1)
```

### Dataset export

```python
df_final = pd.concat([df_num, df_cat_bin_encode, df_cat_multi_encode], axis=1)
```
