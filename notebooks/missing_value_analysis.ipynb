### Helper Function

```python
def fill_one_stage(df, group_col, target_col, method):
    df_fill = df.copy()

    supported_methods = ['mean', 'median', 'mode', 'ffill', 'bfill']

    if method not in supported_methods:
        raise ValueError(f"Method '{method}' not supported. Choose from: {', '.join(supported_methods)}")

    nan_count_initial = df_fill[target_col].isnull().sum()

    if method == 'mode':
        fill_value = df_fill.groupby(group_col)[target_col].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
        df_fill[target_col] = df_fill[target_col].fillna(fill_value)

    elif method in ['mean', 'median']:
        fill_value = df_fill.groupby(group_col)[target_col].transform(method)
        df_fill[target_col] = df_fill[target_col].fillna(fill_value)

    elif method == 'ffill':
        df_fill[target_col] = df_fill.groupby(group_col)[target_col].transform(lambda x: x.ffill().bfill())

    elif method == 'bfill':
        df_fill[target_col] = df_fill.groupby(group_col)[target_col].transform(lambda x: x.bfill().ffill())

    nan_count_after_group = df_fill[target_col].isnull().sum()

    if nan_count_after_group > 0:

        if method == 'mode':
            fill_value_globe = df_fill[target_col].mode().iloc[0] if not df_fill[target_col].mode().empty else np.nan
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'mean':
            fill_value_globe = df_fill[target_col].mean()
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'median':
            fill_value_globe = df_fill[target_col].median()
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'ffill':
            df_fill[target_col].ffill(inplace=True)
            df_fill[target_col].bfill(inplace=True)

        elif method == 'bfill':
            df_fill[target_col].bfill(inplace=True)
            df_fill[target_col].ffill(inplace=True)

    nan_count_final = df_fill[target_col].isnull().sum()

    return df_fill
```

```python
def fill_one_stage_iterative(df, group_col, target_col, method):
    df_fill = df.copy()

    supported_methods = ['mean', 'median', 'mode', 'ffill', 'bfill']

    if method not in supported_methods:
        raise ValueError(f"Method '{method}' not supported. Choose from: {', '.join(supported_methods)}")

    nan_count_initial = df_fill[target_col].isnull().sum()

    def get_fill_value(series, method_str):
        if series.empty or series.isnull().all():
            return np.nan

        if method_str == 'mode':
            modes = series.mode()
            return modes.iloc[0]
        elif method_str == 'mean':
            return series.mean()
        elif method_str == 'median':
            return series.median()

        return np.nan


    for group in df_fill[group_col].unique():
        cond = df_fill[group_col] == group
        df_group = df_fill.loc[cond, target_col]

        if method in ['mean', 'median', 'mode']:
            fill_value = get_fill_value(df_group, method)
            if not pd.isna(fill_value):
                df_fill.loc[cond, target_col] = df_group.fillna(fill_value)
        elif method == 'ffill':
            df_fill.loc[cond, target_col] = df_group.ffill().bfill()
        elif method == 'bfill':
            df_fill.loc[cond, target_col] = df_group.bfill().ffill()

    nan_count_after_group = df_fill[target_col].isnull().sum()

    if nan_count_after_group > 0:

        if method in ['mean', 'median', 'mode']:
            fill_value_globe = get_fill_value(df_fill[target_col], method)
            if not pd.isna(fill_value_globe):
                df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'ffill':
           df_fill[target_col].ffill(inplace=True)
           df_fill[target_col].bfill(inplace=True)

        elif method == 'bfill':
           df_fill[target_col].bfill(inplace=True)
           df_fill[target_col].ffill(inplace=True)

    nan_count_final = df_fill[target_col].isnull().sum()

    return df_fill
```

```python
def fill_two_stage(df, group_col1, group_col2, target_col, method):
    df_fill = df.copy()

    supported_methods = ['mean', 'median', 'mode', 'ffill', 'bfill']
    if method not in supported_methods:
        raise ValueError(f"Method '{method}' not supported. Choose from: {', '.join(supported_methods)}")

    nan_count_initial = df_fill[target_col].isnull().sum()

    if method == 'mode':
        fill_value = df_fill.groupby([group_col1, group_col2])[target_col].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
        df_fill[target_col] = df_fill[target_col].fillna(fill_value)
    elif method in ['mean', 'median']:
        fill_value = df_fill.groupby([group_col1, group_col2])[target_col].transform(method)
        df_fill[target_col] = df_fill[target_col].fillna(fill_value)
    elif method == 'ffill':
        df_fill[target_col] = df_fill.groupby([group_col1, group_col2])[target_col].transform(lambda x: x.ffill().bfill())
    elif method == 'bfill':
        df_fill[target_col] = df_fill.groupby([group_col1, group_col2])[target_col].transform(lambda x: x.bfill().ffill())

    nan_count_level1 = df_fill[target_col].isnull().sum()

    if nan_count_level1 > 0:
        if method == 'mode':
            fill_value_group = df_fill.groupby(group_col1)[target_col].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_group)
        elif method in ['mean', 'median']:
            fill_value_group = df_fill.groupby(group_col1)[target_col].transform(method)
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_group)
        elif method == 'ffill':
            df_fill[target_col] = df_fill.groupby(group_col1)[target_col].transform(lambda x: x.ffill().bfill())
        elif method == 'bfill':
            df_fill[target_col] = df_fill.groupby(group_col1)[target_col].transform(lambda x: x.bfill().ffill())

    nan_count_level2 = df_fill[target_col].isnull().sum()

    if nan_count_level2 > 0:
        if method == 'mode':
            fill_value_globe = df_fill[target_col].mode().iloc[0] if not df_fill[target_col].mode().empty else np.nan
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'mean':
            fill_value_globe = df_fill[target_col].mean()
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'median':
            fill_value_globe = df_fill[target_col].median()
            df_fill[target_col] = df_fill[target_col].fillna(fill_value_globe)

        elif method == 'ffill':
            df_fill[target_col].ffill(inplace=True)
            df_fill[target_col].bfill(inplace=True)

        elif method == 'bfill':
            df_fill[target_col].bfill(inplace=True)
            df_fill[target_col].ffill(inplace=True)

    nan_count_final = df_fill[target_col].isnull().sum()

    return df_fill
```

```python
def fill_two_stage_iterative(df, group_col1, group_col2, target_col, method):
    df_fill = df.copy()

    supported_methods = ['mean', 'median', 'mode', 'ffill', 'bfill']
    if method not in supported_methods:
        raise ValueError(f"Method '{method}' not supported. Choose from: {', '.join(supported_methods)}")

    nan_count_initial = df_fill[target_col].isnull().sum()

    def get_fill_value(series, method_str):

        if series.empty or series.isnull().all():
            return np.nan

        if method_str == 'mode':
            modes = series.mode()
            return modes.iloc[0] if not modes.empty else np.nan
        elif method_str == 'mean':
            return series.mean()
        elif method_str == 'median':
            return series.median()

        return np.nan


    for group1 in df_fill[group_col1].unique():
        for group2 in df_fill[group_col2].unique():
            cond_level1 = (df_fill[group_col1] == group1) & (df_fill[group_col2] == group2)

            if cond_level1.any():
                if method in ['mean', 'median', 'mode']:
                    group_series_level1 = df_fill.loc[cond_level1, target_col]
                    fill_value = get_fill_value(group_series_level1, method)
                    if not pd.isna(fill_value):
                        df_fill.loc[cond_level1, target_col].fillna(fill_value, inplace=True)
                elif method == 'ffill':
                    df_fill.loc[cond_level1, target_col] = df_fill.loc[cond_level1, target_col].ffill().bfill()
                elif method == 'bfill':
                    df_fill.loc[cond_level1, target_col] = df_fill.loc[cond_level1, target_col].bfill().ffill()

    nan_count_level1 = df_fill[target_col].isnull().sum()

    if nan_count_level1 > 0:
        for group1 in df_fill[group_col1].unique():
            cond_level2 = df_fill[group_col1] == group1

            if cond_level2.any():
                if method in ['mean', 'median', 'mode']:
                    group_series_level2 = df_fill.loc[cond_level2, target_col]
                    fill_value_group = get_fill_value(group_series_level2, method)
                    if not pd.isna(group_series_level2):
                        df_fill.loc[cond_level2, target_col].fillna(fill_value_group, inplace=True)
                elif method == 'ffill':
                    df_fill.loc[cond_level2, target_col] = df_fill.loc[cond_level2, target_col].ffill().bfill()
                elif method == 'bfill':
                    df_fill.loc[cond_level2, target_col] = df_fill.loc[cond_level2, target_col].bfill().ffill()

    nan_count_level2 = df_fill[target_col].isnull().sum()

    if nan_count_level2 > 0:
        if method in ['mean', 'median', 'mode']:
            fill_value_globe = get_fill_value(df_fill[target_col], method)
            if not pd.isna(fill_value_globe):
                df_fill[target_col].fillna(fill_value_globe, inplace=True)
        elif method == 'ffill':
            df_fill[target_col].ffill(inplace=True)
            df_fill[target_col].bfill(inplace=True)
        elif method == 'bfill':
            df_fill[target_col].bfill(inplace=True)
            df_fill[target_col].ffill(inplace=True)

    nan_count_final = df_fill[target_col].isnull().sum()

    return df_fill
```
